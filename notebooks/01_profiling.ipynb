```python
for ten, df in bang.items():
    for cot in df.columns:
        if any(k in cot.lower() for k in ['date', 'issued']):
            print(f'{ten}.{cot:<12} kiểu = {df[cot].dtype}    dạng = {df[cot].iloc[0]}')
```

Giá trị mẫu trông như `930101`. Nếu đây là ngày dạng `YYMMDD` thì tách ra phải hợp lý: hai chữ số giữa nằm trong 1–12, hai chữ số cuối nằm trong 1–31.

Viết hàm kiểm tra giả thuyết đó

```python
def kiem_tra_yymmdd(ten, chuoi_so):
    nam   = chuoi_so // 10000
    thang = (chuoi_so // 100) % 100
    ngay  = chuoi_so % 100
    print(f'{ten:<22} nam {nam.min()}-{nam.max()}   '
          f'thang {thang.min()}-{thang.max()}   '
          f'ngay {ngay.min()}-{ngay.max()}')

kiem_tra_yymmdd('account.date', account['date'])
kiem_tra_yymmdd('trans.date',   trans['date'])
kiem_tra_yymmdd('loan.date',    loan['date'])
```

## Nhận xét 3

Mỗi tài khoản sẽ có một người đứng tên chính (OWNER), và một số tài khoản có thêm người được ủy quyền (DISPONENT). Việc số lượng OWNER bằng số lượng tài khoản là hợp lý, vì mỗi tài khoản chỉ có một chủ.

Vì vậy, account_id bị lặp trong disp không phải do dữ liệu bị lỗi, mà là do một tài khoản có thể gắn với hai người với hai vai trò khác nhau.

Điểm này cần lưu ý khi JOIN bảng. Nếu nối trans với disp, những tài khoản có cả OWNER và DISPONENT sẽ làm mỗi giao dịch khớp với 2 dòng. Khi đó, nếu tính số giao dịch hoặc tổng tiền mà không xử lý trước, kết quả rất dễ bị đếm trùng.

```python
# trans JOIN account
kq1 = trans.merge(account, on='account_id')

print('trans:', len(trans))
print('Sau khi join account:', len(kq1))
print('Tăng:', len(kq1) - len(trans))


# trans JOIN disp
kq2 = trans.merge(disp, on='account_id')

print('\ntrans:', len(trans))
print('Sau khi join disp:', len(kq2))
print('Tăng:', len(kq2) - len(trans))
```

Số dòng tăng lên như dự đoán

Kiểm tra số tiền 
balance: đây là số dư *sau mỗi giao dịch*, tức số dư luỹ kế. Cộng toàn bộ cột này là vô nghĩa. Muốn biết tổng số dư danh mục thì lấy số dư của **giao dịch cuối cùng** mỗi tài khoản.

Loại chỉ số này gọi là **semi-additive** — cộng được theo tài khoản nhưng không cộng được theo thời gian. 

```python
so_du_cuoi = (trans
              .sort_values(['account_id', 'date', 'trans_id'])
              .groupby('account_id')['balance']
              .last()
              .reset_index())

so_du_cuoi.head()
```

```python
# Chỉ lấy những dòng là chủ tài khoản
disp_owner = disp[disp['type'] == 'OWNER']
# Tổng số dư đúng
tong_dung = so_du_cuoi['balance'].sum()
print('Tổng số dư đúng:', tong_dung)
# JOIN với toàn bộ disp
merge_day_du = so_du_cuoi.merge(disp, on='account_id')

tong_sai = merge_day_du['balance'].sum()

print('Tổng sau khi JOIN với disp:', tong_sai)

# JOIN chỉ với OWNER
merge_owner = so_du_cuoi.merge(disp_owner, on='account_id')

tong_owner = merge_owner['balance'].sum()

print('Tổng sau khi chỉ lấy OWNER:', tong_owner)
```

Lại quét đều cả 8 bảng. Liệt kê toàn bộ giá trị của những cột có ít giá trị khác nhau.

```python
def quet_phan_loai(ten, df, nguong=12):
    for cot in df.columns:
        n = df[cot].nunique()
        if 1 < n <= nguong:
            print(f'--- {ten}.{cot}   ({n} gia tri)')
            print(df[cot].value_counts(dropna=False).to_string())
            print()
```

Phần tháng vượt quá 12 — phép thử `YYMMDD` thất bại ở đây, trong khi nó đã thoả với ba cột `date` trước đó.

Nhưng `min = 1, max = 62` chưa đủ để kết luận gì. 

```python
thang = (client['birth_number'] // 100) % 100

thang.value_counts().sort_index()
```

Phân bố tách thành **hai cụm rời nhau**, không phải dải liên tục. Đo khoảng cách và kích thước hai cụm.

```python
cum_thap = thang[thang <= 12]
cum_cao  = thang[thang > 12]

print(f'Cum thap: {cum_thap.min()}-{cum_thap.max()}   co {len(cum_thap)} nguoi')
print(f'Cum cao : {cum_cao.min()}-{cum_cao.max()}   co {len(cum_cao)} nguoi')
print()
print('Do lech giua hai cum:', cum_cao.min() - cum_thap.min())
print(f'Ty le: {100 * len(cum_thap) / len(client):.1f}% / {100 * len(cum_cao) / len(client):.1f}%')
```

## Nhận xét 


1. Hai cụm cách nhau **đúng 50** — một số tròn, không phải nhiễu
2. Cụm cao trải **đúng 12 giá trị**, giống hệt cụm thấp — vẫn là 12 tháng
3. Hai cụm **gần bằng nhau về số lượng**, xấp xỉ 50/50


Tra cứu với từ khoá *"Czech birth number encoding"* xác nhận: đây là mã định danh cá nhân của Séc (*rodné číslo*), quy ước cộng 50 vào phần tháng đối với nữ. Một con số mang được cả ngày sinh lẫn giới tính.

**Lưu ý:** dữ liệu chứng minh được *"một trong hai nhóm bị cộng 50"*, nhưng **không** chứng minh được nhóm đó là nữ hay nam. 

**Quyết định: tách `birth_number` thành hai cột** — `birth_date` và `gender`.



# TỔNG KẾT
## Bảng quyết định làm sạch

| # | Vấn đề | Quyết định |
|---|---|---|
| 1 | `district.A12`, `A15` chứa `?` | Để NULL, không điền thay thế |
| 2 | Cột ngày là số nguyên `YYMMDD` | Đổi sang kiểu `DATE` |
| 3 | `disp` có 2 dòng cho 1 tài khoản | Lọc `type = 'OWNER'` |
| 4 | `operation` rỗng 183.114 dòng | Điền nhãn `INTEREST_CREDIT` |
| 5 | `birth_number` mã hoá giới tính | Tách thành `birth_date` + `gender` |

| 7 | `k_symbol` có 2 kiểu rỗng | `NULLIF(TRIM(k_symbol), '')` |
| 8 | Số dư âm | **Giữ nguyên**  |
